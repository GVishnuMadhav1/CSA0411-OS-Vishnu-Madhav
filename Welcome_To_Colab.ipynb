```
import cv2
import matplotlib.pyplot as plt

# Step 1: Upload image
from google.colab import files
uploaded = files.upload()

# Step 2: Read image
# Replace 'input.jpg' with the actual filename uploaded by the user
image = cv2.imread('one-piece-monkey-d-luffy-hd-wallpaper-preview.jpg')

# Check if the image was loaded successfully
if image is None:
    print("Error: Could not load image. Please ensure the filename is correct and the file exists.")
else:
    # Step 3: Convert to grayscale
    gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    # Step 4: Display images
    plt.figure(figsize=(10,5))

    plt.subplot(1,2,1)
    plt.title("Original Image")
    plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    plt.axis('off')

    plt.subplot(1,2,2)
    plt.title("Grayscale Image")
    plt.imshow(gray_image, cmap='gray')
    plt.axis('off')

    plt.show()
```
